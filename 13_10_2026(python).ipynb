```python
"""
Función que retorne el promedio de 4 califiaciones
Criterio: 4% de la primera nota, 12% de la segunda nota,
24% de la tercera nota, y el 60% de la cuarta nota.
"""
def promedio(a, b, c, d):
    return (a*0.04)+(b*0.12)+(c*0.24)+(d*0.6)

pro = promedio(11, 15, 10, 18)
print(pro)

```

```python

"""
Función que la suma total de una lista de "n" elementos
"""
def suma_lista(lista):
    s = 0
    for i in lista:
        s= s + i
    return s

total = suma_lista([10, 5, 1, 40, 30, 25, 14])
print(total)
```

```python

"""
Función que retorne de una lista de "n" elementos, solo
los valores pares, en otra lista.
"""
def pares(lista):
    numeros_pares = []
    for i in lista:
        if (i % 2 == 1):
            numeros_pares.append(i)
    return numeros_pares

Lista = pares([10, 30, 25, 5, 16, 7, 9])
print(Lista)

```

```python
"""
Función que retorno la suma de dos listas de "n" elementos similares en tamaño,
en una terera lista.
Ejemplo:
[10, 5, 10] +
[5, 20, 5]
--------------
[15, 25, 15] <----
"""
def sumar2Listar(Lista1, Lista2):
    L = []
    for i in range(len(Lista1)):
        L.append(Lista1[i] + Lista2[i])
    return L


Lista = sumar2Listar([1,2,3] , [3,2,1])
print(Lista)
```

```python
"""
Función que retorne  la suma de los cuadrados, de una lista
de "n" elementos.

Ejemplo:

[10, 5, 2, 8]

      2     2       2        2
--> 10   + 5   +   2    +   8  = 193
"""
def sumarCuadrados(Lista):
    acum = 0
    for i in Lista:
        acum += (i**2)
    return acum

x = sumarCuadrados([1,2,3,4,5,6,7,8,9,0])
print(x)
```
