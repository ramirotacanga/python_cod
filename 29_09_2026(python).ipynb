```python
# Modulo: miProyecto1.py

# La base de datos

menu = ["1. Agregar Productos",
        "2. Retirar Productos",
        "3. Modificar Productos",
        "4. Listar Productos",
        "5. Aumentar Stock",
        "6. Reducir Stock",
        "7. Salir"]

productos = ["Arroz","Azucar","Fideos","Aceite"]
precios = [3.50, 4.00, 1.50, 11.00]
stock = [25, 20, 40, 30]



# Modulos auxiliares

def listar():
    print()
    print("--------------------------------------")
    print("Listado de Productos en Almacén")
    print("--------------------------------------")
    for i in range(len(productos)): # 0,1,2,3,....
        print(i+1,productos[i],"\t",precios[i],"\t",stock[i])

    print("Fin del listado")
    print()
    salida = input("Presione Enter para continuar....")
    print()

def agregar():
    print()
    print("--------------------------------------")
    print("Agregar nuevos productos en Almacén")
    print("--------------------------------------")
    newProducto = input("Ingrese nuevo producto: ")
    newPrecio = float(input("Ingrese nuevo precio: "))
    newStock = int(input("Ingrese nuevo stock: "))
    productos.append(newProducto)
    precios.append(newPrecio)
    stock.append(newStock)
    print("Productos Agregado....")
    print()
    salida = input("Presione Enter para continuar....")

def retirar():
    print()
    print("--------------------------------------")
    print("Retirar productos en Almacén")
    print("--------------------------------------")
    for i in range(len(productos)): # 0,1,2,3,....
        print(i+1,productos[i],"\t",precios[i],"\t",stock[i])
    print("--------------------------------------")
    op = int(input("Seleccione Producto :"))
    del productos[op-1]
    del precios[op-1]
    del stock[op-1]
    print()
    print("Producto Retirado de Almacen")
    print()
    salida = input("Presione Enter para salir.... ")

def modificar():
    print()
    print("--------------------------------------")
    print("Modificar productos en Almacén")
    print("--------------------------------------")
    for i in range(len(productos)): # 0,1,2,3,....
        print(i+1,productos[i],"\t",precios[i],"\t",stock[i])
    print("--------------------------------------")
    op = int(input("Seleccione Producto :"))
    print("--------------------------------------")
    newProducto = input("Modifique Producto: ")
    newPrecio = float(input("Modifique Precio: "))
    newStock = int(input("Modifique Stock: "))
    productos[op-1] = newProducto
    precios[op-1] = newPrecio
    stock[op-1] = newStock
    print("Producto Modificado con Exito...!!!!")
    print()


def aumentarStock()
    print()
    print("--------------------------------------")
    print("Aumentar Stock productos en Almacén")
    print("--------------------------------------")
    for i in range(len(productos)): # 0,1,2,3,....
        print(i+1,productos[i],"\t",precios[i],"\t",stock[i])
    print("--------------------------------------")
    op = int(input("Seleccione Producto :"))
    cantidad = int(input("Seleccione Producto: "))
    stock [op-1] += cantidad
    print("Producto Acumulado....")
    print()

    def reducirStock()
    print()
    print("--------------------------------------")
    print("Reducir Stock productos en Almacén")
    print("--------------------------------------")
    for i in range(len(productos)): # 0,1,2,3,....
        print(i+1,productos[i],"\t",precios[i],"\t",stock[i])
    print("--------------------------------------")
    op = int(input("Seleccione Producto :"))
    cantidad = int(input("Seleccione Producto: "))
    if cantidad < 0:
      cantidad * -1

    stock [op-1] -= cantidad
    print("Producto Reducido....")
    print()


# Programa Principal
while (True):
    print("--------------------------------------")
    print("Menu Principal del Sistema")
    print("--------------------------------------")
    for i in range(7): # 0,1,2,3,4,5,6
        print(menu[i])
    print("--------------------------------------")
    opcion = int(input("Seleccione Opción: "))
    if (opcion == 1):
        agregar()

    elif (opcion == 2):
        retirar()

    elif (opcion == 3):
        modificar()

    elif (opcion == 4):
        listar()

    elif (opcion == 5):
        pass
    elif (opcion == 6):
        pass
    elif (opcion == 7):
        print()
        print("GRACIAS POR USAR NUESTRA APP !!")
        break
```
