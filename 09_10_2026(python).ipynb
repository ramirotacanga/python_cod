```python
# Modulo: tema7_ejemplo1.py

"""
Funciones con valores de retorno
"""

def leerValor1():
    return float(input("Ingrese Primer valor : "))

def leerValor2():
    return float(input("Ingrese Segundo valor: "))

def suma(a, b):
    return a+b

def resta(a, b):
    return a-b

def multiplica(a, b):
    return a*b

def división(a, b):
    return a/b



# Programa Principal

x = leerValor1()
y = leerValor2()

s = suma(x, y)
r = resta(x, y)
m = multiplica(x, y)
d = división(x, y)

print("-------------------------------------")
print("Operaciones Básicas: ")
print("-------------------------------------")
print("Primer Valor         : ", x)
print("Segundo Valor        : ", y)
print("-------------------------------------")
print("Suma                 : ", s)
print("Resta                : ", r)
print("Multiplicación       : ", m)
print("División             : ", d)
```

```python
# Modulo: tema7_ejemplo2.py

"""
Funciones con valor de retorno

Ingresar nombre y asignatura de un alumno
con 4 calificaciones, obtener el promedio
y determinar si esta o no aprobado.

promedio = (ex1*4%)+(ex2*12%)+(ex3*24%)+(ef*60%)
"""

def leerNombre():
    return input("Ingrese Nombre: ")

def leerAsignatura():
    return input("Ingrese Asignatura: ")

def leerNota1():
    return int(input("Ingrese Nota 1: "))

def leerNota2():
    return int(input("Ingrese Nota 2: "))

def leerNota3():
    return int(input("Ingrese Nota 3: "))

def leerNota4():
    return int(input("Ingrese Nota 4: "))

def promedio(a,b,c,d):
    return (a*0.04)+(b*0.12)+(c*0.24)+(d*0.6)

def estado(prom):
    if (prom >= 12.5):
        return "Aprobado"
    else:
        return "Desaprobado"





# Programa Principal
nombre = leerNombre()
asignatura = leerAsignatura()
n1 = leerNota1()
n2 = leerNota2()
n3 = leerNota3()
n4 = leerNota4()

p = promedio(n1,n2,n3,n4)
e = estado(p)

print("--------------------------------------------")
print("Detalle de Calificaciones IDAT")
print("--------------------------------------------")
print("Nombre                 : ", nombre)
print("Asignatura             : ", asignatura)
print("Calificación 1         : ", n1)
print("Calificación 2         : ", n2)
print("Calificación 3         : ", n3)
print("Calificación 4         : ", n4)
print("--------------------------------------------")
print("Promedio               : ", p)
print("Estado                 : ", e)

```

```python
# Modulo: tema7_ejemplo3.py

"""
Calcular el área de un triangulo.
"""

def leerBase():
  return float(input("Ingrese Base: "))

def leerAltura():
  return float(input("Ingrese Altura: "))

def areaTriangulo(b, h):
  return (b*h)/2.0

#Programa Principal

base = leerBase()
altura = leerAltura()
areat = areaTriangulo(base, altura)

print("--------------------------------------------")
print("Área del Triangulo: ")
print("--------------------------------------------")
print("Valor Base:          : ", base)
print("Valor Altura         : ", altura)
print("--------------------------------------------")
print("Área del Triángulo   : ", areat)
```

```python
# Modulo: tema7_ejemplo4.py
"""
se desea realizar compras, usando los siguientes datos:
1. nombre del cliente
2. nombre del producto
3. precio del producto
4. cantidad a solicitar

Calcular:
- Importe de compra
- Importe de descuento, 10% por compra
- Importe a pagar
- regalos: una tablet si el importe a pagar
           supera los s/.1000.00, en caso
           contrario una USB 64Gb.

"""

def leerNombreC():
  return input("Ingrese Nombre de Cliente: ")

def leerNombreP():
  return input("Ingrese Nombre de Producto: ")

def leerPrecioP():
  return float(input("Ingrese  Precio Producto: "))

def leerCantidadP():
  return int(input("Ingrese Cantidad Solicitada: "))

def importeCompra(pre, can):
  return pre * can

def importeDescuento(ic):
  return ic * 0.1

def importePagar(icomp, idesc):
  return icomp - idesc

def regalos(ipagar):
  if (ipagar > 1000.00):
    return "Una Tablet"
  else:
    return "Una USB 64Gn"

#Programa Principal
nc = leerNombreC()
np = leerNombreP()
pre = leerPrecioP()
can = leerCantidadP()
ic = importeCompra(pre, can)
ides = importeDescuento(ic)
ipagar = importePagar(ic, ides)
regalo = regalos(ipagar)

print("--------------------------------------------")
print("Detalle de Compra: ")
print("--------------------------------------------")
print("Nombre Cliente               :", nc)
print("Nombre del Producto          :", np)
print("Precio del Producto          :", pre)
print("Cantidad del Producto        :", can)
print("--------------------------------------------")
print("Importe de compra            :", ic)
print("Importe de Descuento         :", ides)
print("Importe a Pagar              :",ipagar)
print("Obsequio                     :", regalo)
print("--------------------------------------------")

```
